```python
import datetime
class player:
    def __init__(self,fname,lname,birth_year) :
        self.fname = fname
        self.lname = lname
        self.birth_year = birth_year
        
    def get_age(self):
        now = datetime.datetime.now()
        return now.year - self.birth_year
    
```

```python
class cricker(player):
    def __init__(self,fname,lname,birth_year,team) :
        player.__init__(self,fname,lname,birth_year)
        self.team = team
        self.scores = []
    def add_score(self,score):
        self.scores.append(score)

    def get_average_score(self):
        return sum(self.scores) / len(self.scores)

obj = cricker('virat','kohli',1988,'india')
obj.add_score(100)
obj.add_score(200)
obj.add_score(300)
print(obj.get_average_score())
print(obj.get_age())
```

```python
input_list = [1, 2, 2, 3, 4, 5, 4]
v = {}  # Dictionary to store counts of each element

for i in input_list:
    if i in v:
        v[i] += 1  # Increment the count if the element already exists
    else:
        v[i] = 1   # Initialize the count if the element is seen for the first time

# Print the number of repetitions for each element
print(v)

# You can also count how many elements are repeated (appear more than once)
repeated_count = sum(1 for value in v.values() if value > 1)
print(repeated_count)
```

```python
input_list = [1, 2, 2, 3, 4, 5, 4]
v = {}

for i in input_list:
    if i in v:
        v[i] +=1 # in this step {1:1}
    else:
        v[i] = 1

repeat ={}
for key,value in v.items():
    if value > 1:
        repeat[key]= value
      
print(repeat)

a = list(repeat.keys())
print(a)

```
