```python
def count_consecutive_numbers(lst):
    count = 0
    for i in range(1, len(lst)):
        if lst[i] == lst[i - 1]:
            count += 1
    return count

# Example usage
input_list = [1, 2, 2, 3, 4, 4]
result = count_consecutive_numbers(input_list)
print(result)  # Output: 2
```

```python
# List of numbers
numbers = [7, 1, 2, 2, 8, 3, 4, 3, 5, 4, 1, 6, 9, 11, 12]

# Create an empty dictionary to store the count of each number
count_dict = {}

# Count occurrences of each number
for num in numbers:
    if num in count_dict:
        count_dict[num] += 1
    else:
        count_dict[num] = 1

# Create a list to hold non-repeating numbers
non_repeating_numbers = [num for num in numbers if count_dict[num] == 1]

# Find and print the 3rd non-repeating number
if len(non_repeating_numbers) >= 3:
    print(non_repeating_numbers[2])  # 3rd non-repeating number (index 2)
else:
    print("Less than 3 non-repeating numbers found.")
```

```python
def first_non_repeating_number(numbers):
    # Create a dictionary to store the frequency of each number
    frequency = {}
    
    # Count the occurrences of each number
    for num in numbers:
        if num in frequency:
            frequency[num] += 1
        else:
            frequency[num] = 1
    
    # Find the first number with a frequency of 1
    for num in numbers:
        if frequency[num] == 1:
            return num
    
    return None  # If no non-repeating number is found

# Example usage
numbers = [4, 6,5, 1, 2, 0, 4, 1, 2]
first_non_repeating = first_non_repeating_number(numbers)

if first_non_repeating is not None:
    print(f"The first non-repeating number is: {first_non_repeating}")
else:
    print("No non-repeating number found.")
```

```python
n = [2, 4, 3, 7, 5, 1,6, 9]
# n = [2,  1]
s = 0  # Maximum sum of any increasing subarray
current_sum = n[0]  # Sum of the current increasing subarray

for i in range(1, len(n)):
    if n[i-1] < n[i]:
        current_sum += n[i]  # Add current element to the running sum
    else:
        s = max(s, current_sum)  # Update the max sum if needed
        current_sum = n[i]  # Start a new increasing subarray with the current element

s = max(s, current_sum)  # Final check for the last subarray
print(s)
```
