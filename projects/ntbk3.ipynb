```python
def word_lengths(words):
    if not words:  # Check if the list is empty
        return (0, 0, 0.0)

    lengths = [len(word) for word in words]
    shortest = min(lengths)
    longest = max(lengths)
    average = sum(lengths) / len(lengths)

    return (shortest, longest, average)


words_list = ["apple", "banana", "cherry", "date"]
result = word_lengths(words_list)
print(result)  # Output: (4, 6, 5.0)
```

```python
def sum_of_integers(objects):
    total = 0
    for obj in objects:
        try:
            total += int(obj)
        except (ValueError, TypeError):
            pass  # Ignore objects that can't be converted to integers
    return total


objects_list = [10, "20", "hello", 3.5, [1, 2, 3], "100", None, 50]
result = sum_of_integers(objects_list)
print(result)  # Output: 180
```

```python
def sum_of_numbers(objects):
    total = 0
    for obj in objects:
        if isinstance(obj, list):
            total += sum_of_numbers(obj)  # Recursively sum numbers in nested lists
        else:
            try:
                total += float(
                    obj
                )  # Convert to float to handle both integers and floats
            except (ValueError, TypeError):
                pass  # Ignore objects that can't be converted to floats
    return total


# Example usage:
objects_list = [10, "20", "hello", 3.5, [1, 2, 3], "100", None, 50]
result = sum_of_numbers(objects_list)
print(round(result))  # Output: 189.5
```

```python
def sof(ob):
    total = 0
    for obj in ob:
        if isinstance(obj, list):
            total += sof(obj)

        else:
            try:
                total += float(obj)

            except ValueError:
                pass

    return total


val = ["1", [1, 2, 3], "5"]
res = sof(val)
print(res)
```
